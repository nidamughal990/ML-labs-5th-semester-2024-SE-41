# Question 6:Lists and Tuples
*question*:Difference between a tuple and a list
*Answer*:
A list is a collection of items that can be changed after it is created. We use square brackets [] to create a list.

A tuple is a collection of items that cannot be changed after it is created. We use parentheses () to create a tuple.

Therefore, lists are mutable, while tuples are immutable.


```python
my_list = [1, 2, 3]
my_tuple = (1, 2, 3)

print("Q6 - Original List:", my_list)
print("Q6 - Original Tuple:", my_tuple)

my_list[0] = 100

print("Q6 - Modified List:", my_list)
print("Q6 - Tuple remains unchanged:", my_tuple)
```

# Question 8: Check for dog, ignoring uppercase and lowercase

```python
def find_dog(text):
    return "dog" in text.lower()

print("Q8-test 1:" ,find_dog("Is there a dog here?"))
print("Q8-test 2:",find_dog("I have a DOG."))
print("Q8-test 3:" ,find_dog("I have a cat."))
```

# Question 9: Function: count how many times "dog" occurs in a string (case-insensitive).

```python
def count_dog(text):
    return text.lower().count("dog")

print("Q9-test 1:",count_dog("dog dog DOG"))
print("Q9-test 2:",count_dog("I have a dog and another dog"))
```

# Question 10: caught_speeding(speed, is_birthday):
<= 60 "No Ticket",
 61-80 "Small Ticket", >= 81 "Big Ticket". On a birthday, allow 5 mph more.


```python
def caught_speeding(speed, is_birthday):
    if is_birthday:
        speed = speed - 5

    if speed <= 60:
        return "No Ticket"
    elif speed <= 80:
        return "Small Ticket"
    else:
        return "Big Ticket"

print("Q10-test 1:",caught_speeding(60, False))
print("Q10-test 2:",caught_speeding(75, False))
print("Q10-test 3:",caught_speeding(85, False))
print( "Q10-test 4:",caught_speeding(65, True))
print("Q10-test 5:",caught_speeding(85, True))
```
