```python
# Task 6: Difference between a tuple and a list

# A list is mutable, which means we can change its elements.
# A tuple is immutable, which means we cannot change its elements.

my_list = [1, 2, 3]
my_tuple = (1, 2, 3)

print("List:", my_list)
print("Tuple:", my_tuple)

my_list[0] = 100
print("Modified list:", my_list)
```

```python
# Task 8: Check for dog, ignoring uppercase and lowercase

def find_dog(text):
    return "dog" in text.lower()

print(find_dog("Is there a dog here?"))
print(find_dog("I have a DOG."))
print(find_dog("I have a cat."))
```

```python
# Task 9: Function: count how many times "dog" occurs in a string (case-insensitive).

def count_dog(text):
    return text.lower().count("dog")

print(count_dog("dog dog DOG"))
print(count_dog("I have a dog and another dog"))
```

```python
# Task 10: caught_speeding(speed, is_birthday): <= 60 "No Ticket",
# 61-80 "Small Ticket", >= 81 "Big Ticket". On a birthday, allow 5 mph more.

def caught_speeding(speed, is_birthday):
    if is_birthday:
        speed = speed - 5

    if speed <= 60:
        return "No Ticket"
    elif speed <= 80:
        return "Small Ticket"
    else:
        return "Big Ticket"

print(caught_speeding(60, False))
print(caught_speeding(75, False))
print(caught_speeding(85, False))
print(caught_speeding(65, True))
print(caught_speeding(85, True))
```
